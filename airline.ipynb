```python
# DATA CLEANING AND TRANSFORMATION
```

```python
df = df.drop(columns='Unnamed: 0')
```

```python
df = df.dropna()
```

```python
df['Arrival Delay in Minutes'].isnull().sum()
```

```python
df['Arrival Delay in Minutes'] = df['Arrival Delay in Minutes'].astype(int)
print(df['Arrival Delay in Minutes'].dtype)
```

```python
service_cols = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness'
]

df['Average_Service_Rating'] = df[service_cols].mean(axis=1)
```

```python
df['Average_Service_Rating'].head()
```

```python
df['Average_Service_Rating'] = df['Average_Service_Rating'].round(1)
df['Average_Service_Rating'].head()
```

```python
# AVERAGE RATING OF ALL SERVICES

service_avg = df[service_cols].mean().sort_values()

plt.figure(figsize=(12, 8))

sns.barplot(
    x=service_avg.values,
    y=service_avg.index,
    color = 'plum',
    height = 0.5,
    edgecolor = 'purple'
)

plt.xlabel('Average Rating')
plt.ylabel('Service')
plt.title('Average Rating of Airline Services')
plt.tight_layout()

plt.show()
```

```python
# DEPARTURE DELAY VS ARRIVAL DELAY

plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=df,
    x='Departure Delay in Minutes',
    y='Arrival Delay in Minutes',
    hue='satisfaction',
    alpha=0.8
)

plt.xlabel('Departure Delay (Minutes)')
plt.ylabel('Arrival Delay (Minutes)')
plt.title('Departure Delay vs Arrival Delay')

plt.show()
```

```python
# AVERAGE RATING VS SATISFACTION

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x='satisfaction',
    y='Average_Service_Rating',
    color = 'pink',
    edgecolor='blue',
    linewidth=1
)

plt.xlabel('Satisfaction')
plt.ylabel('Average Service Rating')
plt.title('Average Service Rating vs Passenger Satisfaction')
plt.tight_layout()

plt.show()
```

```python
# SATISFACTION COUNT BY CLASS

plt.figure(figsize = (10,6))

sns.countplot(data = df,
              x = 'Class',
              hue = 'satisfaction',
              palette = 'coolwarm',
              edgecolor = 'black')

plt.title('Satisfaction By Class')
plt.xlabel('Class')
plt.ylabel('Satisfaction')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x='Customer Type',
    hue='satisfaction',
    palette = 'Paired',
    edgecolor='black',
    linewidth=1
)

plt.xlabel('Customer Type')
plt.ylabel('Number of Passengers')
plt.title('Passenger Satisfaction by Customer Type')
plt.tight_layout()

plt.show()
```
